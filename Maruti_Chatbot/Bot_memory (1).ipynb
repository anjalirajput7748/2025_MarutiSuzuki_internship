```python
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
#Step1:Load raw PDF(s)
DATA_PATH="data/"
def load_pdf_files(data):
    loader=DirectoryLoader(data,
                            glob='*.pdf',
                            loader_cls=PyPDFLoader)
    documents=loader.load()
    return documents
documents=load_pdf_files(data=DATA_PATH)
print("Length of PDF pages:",len(documents))
```

```python
#step2:Create Chunks
def create_chunks(extracted_data):
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,
                                                 chunk_overlap=50)
    text_chunks=text_splitter.split_documents(extracted_data)
    return text_chunks
text_chunks=create_chunks(extracted_data=documents)
print("Length of Text Chunks:",len(text_chunks))
```

```python
#Step2:2 connect LLM with FAIIS and Create chain
DB_FAISS_PATH="vectorstore/db_faiss"
CUSTOM_PROMPT_TEMPLATE ="""
Use the pieces of information provided in the context to answer user's question.
if you dont know the answer,just say that you dont know, dont try to make up an answer.
#Dont provide anything out of the given context
can give some creative answer also but only related to the correct information.
give answers in points or paragraph based on type of query.

Context:{context}
Question:{question}

Start the answer directly. No small talk please.
"""
def set_custom_prompt(custom_prompt_template):
    prompt=PromptTemplate(custom_prompt_template,input_variables=["context","question"])
    return prompt



```

```python
# Fix set_custom_prompt to use PromptTemplate correctly
from langchain_core.prompts import PromptTemplate
def set_custom_prompt(custom_prompt_template):
    prompt = PromptTemplate(template=custom_prompt_template, input_variables=["context", "question"])
    return prompt

# Re-run QA chain creation and invocation
from langchain_community.llms import Ollama
from langchain_community.vectorstores import FAISS
from langchain.chains import RetrievalQA

llm = Ollama(model="gemma2:2b", temperature=2)
qa_chain = RetrievalQA.from_chain_type(
    llm =llm,
    chain_type="stuff",
    retriever=db.as_retriever(search_kwargs={"k": 5}),
    return_source_documents=True,
    chain_type_kwargs={"prompt": set_custom_prompt(CUSTOM_PROMPT_TEMPLATE)}
)
# now invoke with a single query
user_query = input("Write Query Here:")
response = qa_chain.invoke({'query': user_query})
print("RESULT:", response["result"])
print("\nSOURCE DOCUMENTS:")
for doc in response["source_documents"]:
    source = doc.metadata.get("source", "Unknown source")
    page = doc.metadata.get("page", "Unknown page")
    print(f"- {source} (Page {page})")

```
